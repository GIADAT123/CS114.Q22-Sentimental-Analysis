```python
import pandas as pd
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, accuracy_score, f1_score, precision_score, recall_score
```

```python


TRAIN_PATH = "/content/drive/MyDrive/CS114/TeamModel/data/train_clean_tokenizedEmoji.csv"
VAL_PATH = "/content/drive/MyDrive/CS114/TeamModel/data/val_clean_tokenizedEmoji.csv"
TEST_PATH = "/content/drive/MyDrive/CS114/TeamModel/data/test_clean_tokenizedEmoji.csv"

TEXT_COL = "text"
LABEL_COL = "status"

train_df = pd.read_csv(TRAIN_PATH)
val_df = pd.read_csv(VAL_PATH)
test_df = pd.read_csv(TEST_PATH)

print("Train:", train_df.shape)
print("Val:", val_df.shape)
print("Test:", test_df.shape)


X_train_text = train_df[TEXT_COL].astype(str)
y_train = train_df[LABEL_COL]

X_val_text = val_df[TEXT_COL].astype(str)
y_val = val_df[LABEL_COL]

X_test_text = test_df[TEXT_COL].astype(str)
y_test = test_df[LABEL_COL]
```

```python
tfidf_lr_model = Pipeline([
    ("tfidf", TfidfVectorizer(
        max_features=50000,
        ngram_range=(1, 2),
        min_df=2
    )),
    ("clf", LogisticRegression(
        max_iter=2000,
        class_weight="balanced",
        random_state=42
    ))
])

tfidf_lr_model.fit(X_train_text, y_train)

y_train_pred = tfidf_lr_model.predict(X_train_text)
y_val_pred = tfidf_lr_model.predict(X_val_text)
y_test_pred = tfidf_lr_model.predict(X_test_text)

splits = [
        ("TRAIN SET", y_train, y_train_pred),
        ("VALIDATION SET", y_val, y_val_pred),
        ("TEST SET", y_test, y_test_pred),
    ]

for name, y_true, y_pred in splits:
        section = f"--- ĐÁNH GIÁ {name} ---\n"
        section += classification_report(y_true, y_pred, digits=4)
        section += "\n" + "=" * 80 + "\n\n"
        print(section)
```
